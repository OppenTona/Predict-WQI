```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.path as mpath

def draw_sankey(ax, nodes, flows, title_years):
    # Cấu hình kích thước
    node_width = 0.5    # Độ rộng của cột (block)
    layer_width = 10    # Khoảng cách giữa các cột mốc thời gian
    vertical_gap = 0.5  # Khoảng cách giữa các block trong cùng 1 cột
    
    # Tính toán vị trí
    node_positions = {} 
    node_usage = {} 
    
    for i, layer in enumerate(nodes):
        # Tính tổng chiều cao của cột để căn giữa
        total_value = sum(n[1] for n in layer)
        total_gap = (len(layer) - 1) * vertical_gap
        total_height = total_value + total_gap
        
        # Bắt đầu vẽ từ tọa độ y để block nằm giữa
        y_cursor = total_height / 2
        
        for j, node in enumerate(layer):
            label, value, color = node
            top = y_cursor
            bottom = y_cursor - value
            x = i * layer_width
            
            # Lưu vị trí để nối dây sau này
            node_positions[(i, label)] = {'x': x, 'top': top, 'bottom': bottom, 'color': color}
            node_usage[(i, label)] = {'out': 0, 'in': 0}
            
            # Vẽ Block (Hình chữ nhật)
            rect = patches.Rectangle((x, bottom), node_width, value, 
                                     linewidth=1, edgecolor='black', facecolor=color, alpha=0.8)
            ax.add_patch(rect)
            
            # Viết nhãn (Label)
            text_x = x + node_width + 0.2
            ax.text(text_x, (top+bottom)/2, label, va='center', fontsize=9, fontweight='normal')
            
            y_cursor = bottom - vertical_gap

    # Vẽ các dòng chảy (Flows)
    for flow in flows:
        l_idx, src_lbl, tgt_lbl, val, color = flow
        
        # Lấy tọa độ block nguồn và đích
        if (l_idx, src_lbl) not in node_positions or (l_idx+1, tgt_lbl) not in node_positions:
            continue
            
        src_node = node_positions[(l_idx, src_lbl)]
        tgt_node = node_positions[(l_idx+1, tgt_lbl)]
        
        # Tính toán vị trí dây nối (stacking)
        src_usage = node_usage[(l_idx, src_lbl)]['out']
        src_y_top = src_node['top'] - src_usage
        src_y_bottom = src_y_top - val
        node_usage[(l_idx, src_lbl)]['out'] += val
        
        tgt_usage = node_usage[(l_idx+1, tgt_lbl)]['in']
        tgt_y_top = tgt_node['top'] - tgt_usage
        tgt_y_bottom = tgt_y_top - val
        node_usage[(l_idx+1, tgt_lbl)]['in'] += val
        
        # Tọa độ đường cong Bezier
        x_start = src_node['x'] + node_width
        x_end = tgt_node['x']
        
        verts = [
            (x_start, src_y_top), # Điểm bắt đầu (trên)
            (x_start + (x_end - x_start)/2, src_y_top), # Điểm điều khiển 1
            (x_start + (x_end - x_start)/2, tgt_y_top), # Điểm điều khiển 2
            (x_end, tgt_y_top), # Điểm kết thúc (trên)
            (x_end, tgt_y_bottom), # Điểm kết thúc (dưới)
            (x_start + (x_end - x_start)/2, tgt_y_bottom), # Điểm điều khiển 3
            (x_start + (x_end - x_start)/2, src_y_bottom), # Điểm điều khiển 4
            (x_start, src_y_bottom), # Điểm bắt đầu (dưới)
            (x_start, src_y_top), # Khép vòng
        ]
        
        codes = [mpath.Path.MOVETO, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CURVE4,
                 mpath.Path.LINETO, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CLOSEPOLY]
        
        path = mpath.Path(verts, codes)
        patch = patches.PathPatch(path, facecolor=color, alpha=0.3, edgecolor='none')
        ax.add_patch(patch)

    # Thêm tiêu đề năm
    for i, year in enumerate(title_years):
        ax.text(i * layer_width + node_width/2, 
                max([n['top'] for k,n in node_positions.items()]) + 1, 
                year, ha='center', fontsize=12, fontweight='bold')

    ax.axis('off')
    ax.autoscale_view()

# --- DỮ LIỆU ĐẦU VÀO ---
# Định nghĩa màu sắc
c_red = '#E15759'
c_blue = '#4E79A7'
c_orange = '#F28E2B'
c_green = '#59A14F'
c_pink = '#FF9DA7'
c_purple = '#B07AA1'
c_teal = '#76B7B2'
c_gray = '#BAB0AC'
c_yellow = '#EDC948'
c_brown = '#8C564B'
c_light_blue = '#A0CBE8'
c_stream = '#B5EAD7' # Màu xanh nhạt cho streamflow

# Cấu trúc dữ liệu Nodes: [Layer 1, Layer 2, Layer 3, Layer 4]
# Mỗi Node gồm: (Tên hiển thị, Độ cao/Giá trị, Màu sắc)
nodes = [
    # 2005-2007
    [('AI', 1, c_red), ('ANN', 1, c_blue)],
    
    # 2008-2014
    [('AI', 1, c_orange), ('ANN', 1, c_red), ('Surface Water Quality', 1, c_pink), 
     ('ML', 1, c_blue), ('Decision Support Techniques', 1, c_purple), ('Support Vector Machine', 1, c_green)],
     
    # 2015-2023
    [('Random Forest', 5, c_green), ('ML', 2, c_red), 
     ('Decision Support Techniques', 1, c_gray), ('Geographic Information Systems', 1, c_light_blue),
     ('SHAP', 1, c_teal), ('Surface water quality', 1, c_pink), ('ARIMA', 1, c_yellow),
     ('Feature Selection', 1, c_brown), ('CNN', 1, c_purple), ('Deep learning', 1, c_blue),
     ('Streamflow prediction', 1, c_stream)],
     
    # 2024-2024
    [('Water Quality', 5, c_blue), ('Extreme Gradient Boosting', 1, c_pink), ('MLP', 1, c_stream),
     ('WQP', 1, c_purple), ('River water quality', 1, c_pink), ('SHAP', 1, c_brown),
     ('Geographic Information Systems', 1, c_orange), ('ML', 1, c_red), ('ANN', 1, c_purple),
     ('WQI', 1, c_green), ('Hybrid model', 1, c_gray), ('Groundwater quality', 1, c_orange),
     ('Ensemble learning models', 1, c_teal), ('Deep learning model', 1, c_gray)]
]

# Cấu trúc dữ liệu Flows: (Index cột nguồn, Tên nguồn, Tên đích, Giá trị, Màu sắc)
# Index cột: 0 (2005), 1 (2008), 2 (2015)
flows = [
    # Từ cột 0 (2005-2007) sang cột 1
    (0, 'AI', 'AI', 1, c_gray),
    (0, 'ANN', 'ANN', 1, c_gray),
    
    # Từ cột 1 (2008-2014) sang cột 2
    (1, 'AI', 'ML', 0.5, c_gray),
    (1, 'ANN', 'ML', 0.5, c_gray),
    (1, 'Surface Water Quality', 'ARIMA', 1, c_gray),
    (1, 'ML', 'Decision Support Techniques', 0.5, c_gray),
    (1, 'ML', 'Deep Learning', 1, c_gray),
    (1, 'Decision Support Techniques', 'Geographic Information Systems', 1, c_gray),
    (1, 'Support Vector Machine', 'Random Forest', 1, c_gray),
    
    # Từ cột 2 (2015-2023) sang cột 3
    (2, 'Random Forest', 'Extreme Gradient Boosting', 1, c_gray),
    (2, 'Random Forest', 'MLP', 1, c_gray),
    (2, 'Random Forest', 'WQP', 1, c_gray),
    (2, 'Random Forest', 'River Water Quality', 1, c_gray),
    (2, 'Random Forest', 'ML', 1, c_gray),
    (2, 'Random Forest', 'ANN', 1, c_gray),
    (2, 'Random Forest', 'WQI', 1, c_gray),
    (2, 'Random Forest', 'Hybrid Model', 1, c_gray),
    (2, 'Random Forest', 'Groundwater Quality', 1, c_gray),
    (2, 'Random Forest', 'Esemble Learning Models', 1, c_gray),
    (2, 'ML', 'Water Quality', 1, c_gray),
    (2, 'ML', 'ML', 1, c_gray),
    (2, 'Decision Support Techniques', 'Water Quality', 1, c_gray),
    (2, 'Geographic Information Systems', 'Geographic Information Systems', 1, c_gray),
    (2, 'SHAP', 'SHAP', 1, c_gray),
    (2, 'Surface water quality', 'River water quality', 1, c_gray),
    (2, 'ARIMA', 'Hybrid model', 1, c_gray),
    (2, 'Feature Selection', 'WQI', 1, c_gray),
    (2, 'CNN', 'Deep learning model', 1, c_gray),
    (2, 'Deep learning', 'Ensemble learning models', 1, c_gray),
    (2, 'Streamflow prediction', 'Water Quality', 1, c_gray),
]

# Thực thi vẽ
fig, ax = plt.subplots(figsize=(16, 10))
draw_sankey(ax, nodes, flows, ['2005–2007', '2008–2014', '2015–2023', '2024–2024'])
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.path as mpath

# --- HÀM TÍNH TOÁN CHIỀU CAO NODE TỰ ĐỘNG ---
def calculate_node_heights(nodes_input, flows_input):
    """
    Tính chiều cao của mỗi node dựa trên tổng flow vào/ra lớn nhất.
    """
    node_in_flux = {}
    node_out_flux = {}
    
    # 1. Cộng tổng flow vào và ra cho từng node
    for layer_idx, src, tgt, value, color in flows_input:
        # Tính luồng ra từ nguồn (Source)
        src_key = (layer_idx, src)
        node_out_flux[src_key] = node_out_flux.get(src_key, 0) + value
        
        # Tính luồng vào đích (Target)
        tgt_key = (layer_idx + 1, tgt)
        node_in_flux[tgt_key] = node_in_flux.get(tgt_key, 0) + value

    # 2. Cập nhật lại danh sách nodes với giá trị mới
    new_nodes = []
    for i, layer in enumerate(nodes_input):
        new_layer = []
        for label, _, color in layer: # Giá trị cũ (_) bị bỏ qua
            val_in = node_in_flux.get((i, label), 0)
            val_out = node_out_flux.get((i, label), 0)
            
            # Chiều cao node = Max(tổng vào, tổng ra)
            # Nếu node cô lập (không có dây) thì cho chiều cao nhỏ mặc định là 0.1
            height = max(val_in, val_out)
            if height == 0: height = 0.1 
            
            new_layer.append((label, height, color))
        new_nodes.append(new_layer)
    return new_nodes

# --- HÀM VẼ ---
def draw_sankey(ax, nodes, flows, title_years):
    # Tự động tính lại chiều cao nodes trước khi vẽ
    nodes = calculate_node_heights(nodes, flows)

    # Cấu hình
    node_width = 0.5
    layer_width = 10
    vertical_gap = 0.3 # Giảm khoảng cách để trông chặt chẽ hơn
    
    node_positions = {} 
    node_usage = {} # Theo dõi vị trí dây đã vẽ để stack chồng lên nhau
    
    # 1. Vẽ Nodes
    for i, layer in enumerate(nodes):
        total_value = sum(n[1] for n in layer)
        total_gap = (len(layer) - 1) * vertical_gap
        total_height = total_value + total_gap
        
        y_cursor = total_height / 2
        
        for j, node in enumerate(layer):
            label, value, color = node
            top = y_cursor
            bottom = y_cursor - value
            x = i * layer_width
            
            node_positions[(i, label)] = {'x': x, 'top': top, 'bottom': bottom, 'color': color}
            node_usage[(i, label)] = {'out': 0, 'in': 0} # Reset bộ đếm usage
            
            # Vẽ Block
            rect = patches.Rectangle((x, bottom), node_width, value, 
                                     linewidth=1, edgecolor='black', facecolor=color, alpha=0.85)
            ax.add_patch(rect)
            
            # Label
            text_x = x + node_width + 0.2
            ax.text(text_x, (top+bottom)/2, label, va='center', fontsize=9)
            
            y_cursor = bottom - vertical_gap

    # 2. Vẽ Flows
    for flow in flows:
        l_idx, src_lbl, tgt_lbl, val, color = flow
        
        if (l_idx, src_lbl) not in node_positions or (l_idx+1, tgt_lbl) not in node_positions:
            continue
            
        src_node = node_positions[(l_idx, src_lbl)]
        tgt_node = node_positions[(l_idx+1, tgt_lbl)]
        
        # Tính toán tọa độ Y bắt đầu (Source)
        # Flow đi ra sẽ xếp từ trên xuống dưới
        src_usage = node_usage[(l_idx, src_lbl)]['out']
        src_y_top = src_node['top'] - src_usage
        src_y_bottom = src_y_top - val
        node_usage[(l_idx, src_lbl)]['out'] += val
        
        # Tính toán tọa độ Y kết thúc (Target)
        # Flow đi vào cũng xếp từ trên xuống dưới
        tgt_usage = node_usage[(l_idx+1, tgt_lbl)]['in']
        tgt_y_top = tgt_node['top'] - tgt_usage
        tgt_y_bottom = tgt_y_top - val
        node_usage[(l_idx+1, tgt_lbl)]['in'] += val
        
        # Vẽ dây cong
        x_start = src_node['x'] + node_width
        x_end = tgt_node['x']
        
        verts = [
            (x_start, src_y_top),
            (x_start + (x_end - x_start)/2, src_y_top),
            (x_start + (x_end - x_start)/2, tgt_y_top),
            (x_end, tgt_y_top),
            (x_end, tgt_y_bottom),
            (x_start + (x_end - x_start)/2, tgt_y_bottom),
            (x_start + (x_end - x_start)/2, src_y_bottom),
            (x_start, src_y_bottom),
            (x_start, src_y_top),
        ]
        
        codes = [mpath.Path.MOVETO, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CURVE4,
                 mpath.Path.LINETO, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CURVE4, mpath.Path.CLOSEPOLY]
        
        path = mpath.Path(verts, codes)
        patch = patches.PathPatch(path, facecolor=color, alpha=0.4, edgecolor='none') # Alpha thấp hơn để nhìn xuyên
        ax.add_patch(patch)

    # Tiêu đề năm
    for i, year in enumerate(title_years):
        ax.text(i * layer_width + node_width/2, 
                max([n['top'] for k,n in node_positions.items()]) + 1, 
                year, ha='center', fontsize=12, fontweight='bold')

    ax.axis('off')
    ax.autoscale_view()

# --- DỮ LIỆU (Giữ nguyên cấu trúc, giá trị size trong nodes có thể để bằng 0 vì sẽ tự tính lại) ---
c_red = '#E15759'
c_blue = '#4E79A7'
c_orange = '#F28E2B'
c_green = '#59A14F'
c_pink = '#FF9DA7'
c_purple = '#B07AA1'
c_teal = '#76B7B2'
c_gray = '#BAB0AC'
c_yellow = '#EDC948'
c_brown = '#8C564B'
c_light_blue = '#A0CBE8'
c_stream = '#B5EAD7'

# Nodes: Giá trị số ở giữa (ví dụ số 1) giờ đây CHỈ LÀ GIẢ, code sẽ tự ghi đè
nodes = [
    [('AI', 0, c_red), ('ANN', 0, c_blue)], # 2005-2007
    
    [('AI', 0, c_orange), ('ANN', 0, c_red), ('Surface Water Quality', 0, c_pink), 
     ('ML', 0, c_blue), ('Decision Support Techniques', 0, c_purple), ('Support Vector Machine', 0, c_green)], # 2008-2014
     
    [('Random Forest', 0, c_green), ('ML', 0, c_red), 
     ('Decision Support Techniques', 0, c_gray), ('Geographic Information Systems', 0, c_light_blue),
     ('SHAP', 0, c_teal), ('Surface water quality', 0, c_pink), ('ARIMA', 0, c_yellow),
     ('Feature Selection', 0, c_brown), ('CNN', 0, c_purple), ('Deep learning', 0, c_blue),
     ('Streamflow prediction', 0, c_stream)], # 2015-2023
     
    [('Water Quality', 0, c_blue), ('Extreme Gradient Boosting', 0, c_pink), ('MLP', 0, c_stream),
     ('WQP', 0, c_purple), ('River water quality', 0, c_pink), ('SHAP', 0, c_brown),
     ('Geographic Information Systems', 0, c_orange), ('ML', 0, c_red), ('ANN', 0, c_purple),
     ('WQI', 0, c_green), ('Hybrid model', 0, c_gray), ('Groundwater quality', 0, c_orange),
     ('Ensemble learning models', 0, c_teal), ('Deep learning model', 0, c_gray)] # 2024
]

# Flows: Đây là phần quan trọng nhất, giá trị ở đây quyết định độ to nhỏ
flows = [
    # 2005 -> 2008
    (0, 'AI', 'AI', 1.0, c_gray),
    (0, 'ANN', 'ANN', 1.0, c_gray),
    
    # 2008 -> 2015
    (1, 'AI', 'Random Forest', 0.6, c_gray),
    (1, 'AI', 'ML', 0.4, c_gray),
    (1, 'ANN', 'Random Forest', 0.5, c_gray),
    (1, 'Surface Water Quality', 'Surface water quality', 1.0, c_gray),
    (1, 'ML', 'ML', 1.0, c_gray),
    (1, 'Decision Support Techniques', 'Decision Support Techniques', 0.8, c_gray),
    (1, 'Decision Support Techniques', 'Water Quality', 0.2, c_gray), # Thêm nhánh nhỏ để cân bằng
    (1, 'Support Vector Machine', 'Random Forest', 1.0, c_gray),
    
    # 2015 -> 2024 (Chia nhỏ dòng chảy để tạo độ phức tạp)
    (2, 'Random Forest', 'Water Quality', 2.0, c_gray),
    (2, 'Random Forest', 'Extreme Gradient Boosting', 0.5, c_gray),
    (2, 'Random Forest', 'MLP', 0.5, c_gray),
    (2, 'Random Forest', 'WQP', 0.5, c_gray),
    
    (2, 'ML', 'ML', 1.0, c_gray),
    (2, 'ML', 'ANN', 0.8, c_gray),
    
    (2, 'Decision Support Techniques', 'Water Quality', 0.8, c_gray),
    (2, 'Geographic Information Systems', 'Geographic Information Systems', 0.8, c_gray),
    (2, 'SHAP', 'SHAP', 0.8, c_gray),
    (2, 'Surface water quality', 'River water quality', 1.0, c_gray),
    (2, 'ARIMA', 'Hybrid model', 0.8, c_gray),
    (2, 'Feature Selection', 'WQI', 0.8, c_gray),
    (2, 'CNN', 'Deep learning model', 0.5, c_gray),
    (2, 'Deep learning', 'Ensemble learning models', 0.8, c_gray),
    (2, 'Streamflow prediction', 'Water Quality', 0.8, c_gray),
]

# Thực thi vẽ
fig, ax = plt.subplots(figsize=(16, 10))
draw_sankey(ax, nodes, flows, ['2005–2007', '2008–2014', '2015–2023', '2024–2024'])
plt.tight_layout()
plt.show()
```
